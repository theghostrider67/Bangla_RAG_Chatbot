```python
import time
from urllib.parse import quote, unquote
from bs4 import BeautifulSoup
import requests

baseUrl = "https://bn.wikisource.org"
apiUrl = "https://bn.wikisource.org/w/api.php"
bookTitle = "শ্রীকান্ত (প্রথম পর্ব)"
headers = {
    "User-Agent": "BanglaRAGStudentBot/1.0 (https://bn.wikisource.org; student@aiub.edu)"
}


def getChapterLinksFromApi(title):
  params = {
      "action": "query",
      "list": "prefixsearch",
      "pssearch": title + "/",
      "pslimit": 50,
      "format": "json",
  }
  data = requests.get(apiUrl, params=params, headers=headers).json()
  chapterLinks = []
  for item in data["query"]["prefixsearch"]:
    pageTitle = item["title"]
    if pageTitle.startswith(title + "/"):
      chapterName = pageTitle.split("/", 1)[1]
      fullUrl = f"{baseUrl}/wiki/{quote(pageTitle.replace(' ', '_'))}"
      chapterLinks.append({"chapter": chapterName, "url": fullUrl})
  return chapterLinks


def scrapeChapterText(chapterUrl):
  response = requests.get(chapterUrl, headers=headers)
  response.raise_for_status()
  soup = BeautifulSoup(response.text, "html.parser")
  contentDiv = soup.find("div", class_="mw-parser-output")

  for unwanted in contentDiv.select(
      ".ws-noexport, .noprint, .mw-editsection, table, style, script, sup"
  ):
    unwanted.decompose()

  paragraphs = [
      p.get_text(" ", strip=True)
      for p in contentDiv.find_all("p")
      if p.get_text(strip=True)
  ]
  if not paragraphs:
    text = contentDiv.get_text("\n", strip=True)
  else:
    text = "\n\n".join(paragraphs)
  return text


chapters = getChapterLinksFromApi(bookTitle)
print(f"Total chapters found: {len(chapters)}")

bookData = []
for ch in chapters:
  text = scrapeChapterText(ch["url"])
  bookData.append({
      "chapter": ch["chapter"],
      "url": unquote(ch["url"]),
      "text": text,
      "charCount": len(text),
  })
  print(f"Scraped: {ch['chapter']} ({len(text)} chars)")
  time.sleep(0.5)
```

```python
import re


def scrapeChapterText(chapterUrl):
  response = requests.get(chapterUrl, headers=headers)
  response.raise_for_status()
  soup = BeautifulSoup(response.text, "html.parser")
  contentDiv = soup.find("div", class_="prp-pages-output") or soup.find(
      "div", class_="mw-parser-output"
  )

  for unwanted in contentDiv.select(
      ".mw-editsection, style, script, sup, .pagenum, .ws-noexport, .noprint"
  ):
    unwanted.decompose()

  rawText = contentDiv.get_text("\n", strip=True)
  cleanText = re.sub(r"\n{3,}", "\n\n", rawText).strip()
  return cleanText


bookData = []
for ch in chapters:
  text = scrapeChapterText(ch["url"])
  bookData.append({
      "chapter": ch["chapter"],
      "url": unquote(ch["url"]),
      "text": text,
      "charCount": len(text),
  })
  print(f"Scraped: {ch['chapter']} ({len(text)} chars)")
  time.sleep(0.5)

print(
    f"\nSample text from {bookData[0]['chapter']}:\n{bookData[0]['text'][:300]}"
)
```

```python
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

rawDocs = [
    Document(
        page_content=item["text"],
        metadata={"chapter": item["chapter"], "source": item["url"]},
    )
    for item in bookData
]

textSplitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150,
    separators=["\n\n", "\n", "।", "?", "!", " ", ""],
)

chunks = textSplitter.split_documents(rawDocs)
print(f"Total chunks created: {len(chunks)}")
print(f"Sample chunk metadata: {chunks[0].metadata}")

embeddingModel = HuggingFaceEmbeddings(
    model_name="BAAI/bge-m3",
    model_kwargs={"device": "cuda"},
    encode_kwargs={"normalize_embeddings": True},
)

vectorStore = FAISS.from_documents(chunks, embeddingModel)
vectorStore.save_local("srikanta_faiss_index")
print("FAISS vector store created and saved successfully.")
```

```python
evalQueries = [
    {
        "query": "শ্রীকান্তের ছোটবেলার বন্ধু ইন্দ্রনাথের সাথে কীভাবে পরিচয় হয়েছিল?",
        "expectedChapter": "দ্বিতীয় পরিচ্ছেদ",
    },
    {
        "query": "ইন্দ্রনাথ নৌকা নিয়ে অন্ধকার রাতে কোথায় যেত?",
        "expectedChapter": "তৃতীয় পরিচ্ছেদ",
    },
    {
        "query": "শাহজীর কুঠিতে অন্নদাদিদির সাথে শ্রীকান্তের দেখা হওয়ার ঘটনা কী?",
        "expectedChapter": "পঞ্চম পরিচ্ছেদ",
    },
    {
        "query": "সাপুড়ে শাহজী ও অন্নদাদিদির সংসারের অবস্থা কেমন ছিল?",
        "expectedChapter": "ষষ্ঠ পরিচ্ছেদ",
    },
    {
        "query": "রাজলক্ষ্মী বা পিয়ারী বাঈজীর সাথে শ্রীকান্তের কীভাবে দেখা হলো?",
        "expectedChapter": "অষ্টম পরিচ্ছেদ",
    },
    {
        "query": "রাজলক্ষ্মী শ্রীকান্তের অসুস্থতার সময় কীভাবে সেবা করেছিল?",
        "expectedChapter": "একাদশ পরিচ্ছেদ",
    },
]


def evaluateChunkingStrategy(chunkSize, chunkOverlap, k=3):
  splitter = RecursiveCharacterTextSplitter(
      chunk_size=chunkSize,
      chunk_overlap=chunkOverlap,
      separators=["\n\n", "\n", "।", "?", "!", " ", ""],
  )
  strategyChunks = splitter.split_documents(rawDocs)
  tempDb = FAISS.from_documents(strategyChunks, embeddingModel)
  retriever = tempDb.as_retriever(search_kwargs={"k": k})

  hits = 0
  for item in evalQueries:
    retrievedDocs = retriever.invoke(item["query"])
    retrievedChapters = [doc.metadata["chapter"] for doc in retrievedDocs]
    if item["expectedChapter"] in retrievedChapters:
      hits += 1

  hitRate = (hits / len(evalQueries)) * 100
  return len(strategyChunks), hits, hitRate


chunksA, hitsA, rateA = evaluateChunkingStrategy(800, 150, k=3)
chunksB, hitsB, rateB = evaluateChunkingStrategy(300, 30, k=3)

print(
    f"Strategy A (chunk_size=800, overlap=150) -> Chunks: {chunksA} | Hit-Rate"
    f" @3: {hitsA}/{len(evalQueries)} ({rateA:.1f}%)"
)
print(
    f"Strategy B (chunk_size=300, overlap=30)  -> Chunks: {chunksB} | Hit-Rate"
    f" @3: {hitsB}/{len(evalQueries)} ({rateB:.1f}%)"
)
```

```python
import json
import os
import gradio as gr
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = "PASTE_YOUR_GROQ_API_KEY_HERE"

retriever = vectorStore.as_retriever(
    search_type="similarity", search_kwargs={"k": 4}
)

llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0.1)

systemPrompt = """You are a strict Bangla literary assistant answering questions solely from Sarat Chandra Chattopadhyay's novel 'শ্রীকান্ত (প্রথম পর্ব)'.

Rules:
1. Answer ONLY using the provided context below.
2. Every single answer MUST cite the exact chapter name(s) (e.g., [সূত্র: প্রথম পরিচ্ছেদ]) where the information came from.
3. If the answer cannot be found in the provided context, strictly reply: "দুঃখিত, এই তথ্যটি 'শ্রীকান্ত (প্রথম পর্ব)' বইয়ের প্রদত্ত অংশে নেই।" — DO NOT guess or use outside knowledge.
4. Respond clearly in Bengali (Bangla).

Context:
{context}"""

promptTemplate = ChatPromptTemplate.from_messages([
    ("system", systemPrompt),
    ("human", "{question}"),
])


def formatDocsWithCitations(docs):
  formatted = []
  for doc in docs:
    chapter = doc.metadata.get("chapter", "অজানা পরিচ্ছেদ")
    formatted.append(f"[পরিচ্ছেদ: {chapter}]\n{doc.page_content}")
  return "\n\n---\n\n".join(formatted)


ragChain = promptTemplate | llm | StrOutputParser()


def askChatbot(question, history):
  if not question.strip():
    return ""
  retrievedDocs = retriever.invoke(question)
  contextText = formatDocsWithCitations(retrievedDocs)
  answer = ragChain.invoke({"context": contextText, "question": question})
  citedChapters = list(
      dict.fromkeys([doc.metadata["chapter"] for doc in retrievedDocs])
  )
  citationFooter = (
      "\n\n📚 **তথ্যসূত্র (Retrieved Chapters):** " + ", ".join(citedChapters)
  )
  return answer + citationFooter


testQuestions = [
    {
        "id": 1,
        "question": "শ্রীকান্ত নিজের জীবনকে কীসের সাথে তুলনা করেছে?",
        "expectedAnswer": (
            "ছোটবেলা থেকেই শ্রীকান্ত নিজের জীবনকে একটা মস্ত 'ছি-ছি-ছি' বলে মনে"
            " করত।"
        ),
        "expectedChapter": "প্রথম পরিচ্ছেদ",
    },
    {
        "id": 2,
        "question": (
            "ইন্দ্রনাথ কীভাবে শ্রীকান্তকে বখাটে ছেলেদের হাত থেকে রক্ষা করেছিল?"
        ),
        "expectedAnswer": (
            "ফুটবল খেলার মাঠে নূতনদার দলের ছেলেরা শ্রীকান্তকে মারতে এলে"
            " ইন্দ্রনাথ একাই তাদের প্রতিহত করে শ্রীকান্তকে বাঁচিয়েছিল।"
        ),
        "expectedChapter": "দ্বিতীয় পরিচ্ছেদ",
    },
    {
        "id": 3,
        "question": "ইন্দ্রনাথ গভীর রাতে ডিঙি নৌকা নিয়ে গঙ্গায় কেন যেত?",
        "expectedAnswer": (
            "ইন্দ্রনাথ জেলেদের জাল থেকে মাছ চুরি করে গরিব মানুষদের সাহায্য করার"
            " জন্য রাতের অন্ধকারে ডিঙি নিয়ে বের হতো।"
        ),
        "expectedChapter": "তৃতীয় পরিচ্ছেদ",
    },
    {
        "id": 4,
        "question": "ইন্দ্রনাথের নৌকায় শ্রীকান্ত কীসের মাংস রান্না করে খেয়েছিল?",
        "expectedAnswer": (
            "চরের নির্জন স্থানে ইন্দ্রনাথ ও শ্রীকান্ত কচ্ছপের ডিম ও মাংস রান্না"
            " করে খেয়েছিল।"
        ),
        "expectedChapter": "চতুর্থ পরিচ্ছেদ",
    },
    {
        "id": 5,
        "question": "সাপুড়ে শাহজীর স্ত্রীর নাম কী এবং তার পরিচয় কী ছিল?",
        "expectedAnswer": (
            "সাপুড়ে শাহজীর স্ত্রীর নাম অন্নদাদিদি; তিনি ছিলেন শ্রীকান্তের"
            " মেজদার শ্বশুরবাড়ির আত্মীয়া ও অত্যন্ত সতী-সাধ্বী নারী।"
        ),
        "expectedChapter": "পঞ্চম পরিচ্ছেদ",
    },
    {
        "id": 6,
        "question": (
            "অন্নদাদিদি কীভাবে সাপুড়ে শাহজীর সংসারের অভাব মেটানোর চেষ্টা"
            " করতেন?"
        ),
        "expectedAnswer": (
            "অন্নদাদিদি গোপনে মানুষের বাড়ির কাজ করে এবং কষ্ট সহ্য করে সংসার"
            " চালাতেন ও স্বামীর সেবা করতেন।"
        ),
        "expectedChapter": "ষষ্ঠ পরিচ্ছেদ",
    },
    {
        "id": 7,
        "question": "শ্রীকান্ত ও ইন্দ্রনাথ ছিনাথ বহুরূপীকে দেখে কেন ভয় পেয়েছিল?",
        "expectedAnswer": (
            "রাতের অন্ধকারে বাঘের ছদ্মবেশে ছিনাথ বহুরূপী বাগানের ঝোপের আড়াল"
            " থেকে গর্জন করে উঠলে তারা সত্যিকারের বাঘ ভেবে ভয় পেয়েছিল।"
        ),
        "expectedChapter": "সপ্তম পরিচ্ছেদ",
    },
    {
        "id": 8,
        "question": (
            "কুমার সাহেবের শিকারের তাঁবুতে কোন বাঈজী গান গাইতে এসেছিলেন?"
        ),
        "expectedAnswer": (
            "কুমার সাহেবের শিকারের তাঁবুতে পাটনার বিখ্যাত পিয়ারী বাঈজী"
            " (রাজলক্ষ্মী) গান গাইতে এসেছিলেন।"
        ),
        "expectedChapter": "অষ্টম পরিচ্ছেদ",
    },
    {
        "id": 9,
        "question": "পিয়ারী বাঈজীর আসল নাম কী এবং ছোটবেলায় তার কী হয়েছিল?",
        "expectedAnswer": (
            "পিয়ারী বাঈজীর আসল নাম রাজলক্ষ্মী; ছোটবেলায় তার মা তাকে কাশীতে"
            " বিক্রি করে দিয়েছিলেন এবং শ্রীকান্তের সাথে তার বাল্যকালের পরিচয়"
            " ছিল।"
        ),
        "expectedChapter": "নবম পরিচ্ছেদ",
    },
    {
        "id": 10,
        "question": (
            "শ্রীকান্তের প্রথম পর্বে রবীন্দ্রনাথ ঠাকুরের নোবেল পুরস্কার পাওয়ার"
            " কথা কোথায় আছে? (Out-of-book test)"
        ),
        "expectedAnswer": (
            "দুঃখিত, এই তথ্যটি 'শ্রীকান্ত (প্রথম পর্ব)' বইয়ের প্রদত্ত অংশে"
            " নেই।"
        ),
        "expectedChapter": "None (Not in book)",
    },
]

with open("test_questions.json", "w", encoding="utf-8") as f:
  json.dump(testQuestions, f, ensure_ascii=False, indent=2)

demo = gr.ChatInterface(
    fn=askChatbot,
    title="📚 শ্রীকান্ত (প্রথম পর্ব) — Bangla RAG Knowledge Base Chatbot",
    description=(
        "শরৎচন্দ্র চট্টোপাধ্যায়ের 'শ্রীকান্ত (প্রথম পর্ব)' উপন্যাস থেকে"
        " যেকোনো প্রশ্ন করুন। প্রতিটি উত্তরের সাথে অধ্যায় (পরিচ্ছেদ) উল্লেখ"
        " থাকবে।"
    ),
    examples=[
        "শ্রীকান্ত নিজের জীবনকে কীসের সাথে তুলনা করেছে?",
        "ইন্দ্রনাথ কীভাবে শ্রীকান্তকে বখাটে ছেলেদের হাত থেকে রক্ষা করেছিল?",
        "পিয়ারী বাঈজীর আসল নাম কী?",
        "শ্রীকান্ত উপন্যাসে রবীন্দ্রনাথ ঠাকুরের নোবেল পুরস্কারের কথা কী লেখা আছে?",
    ],
)

demo.launch(share=True)
```
